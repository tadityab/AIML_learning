### ❓ Q38. Iterate over rows using `iterrows()`.
### ❓ Q39. Iterate using `itertuples()`.
### ❓ Q40. Avoid loops using vectorization.

```python
import time
df_iter = df_main[["Name","Dept","Score"]].head(4).copy()

# Q38 — iterrows()
print("Q38 — iterrows() → (index, Series):")
for idx, row in df_iter.iterrows():
    print(f"  [{idx}] {row['Name']} | {row['Dept']} | {row['Score']}")

# Q39 — itertuples() — faster, named tuples
print("\nQ39 — itertuples() → NamedTuple:")
for row in df_iter.itertuples():
    print(f"  Index={row.Index} | {row.Name} | {row.Score}")

# Q40 — Vectorization is MUCH faster
n = 100_000
big = pd.DataFrame({"val": np.random.rand(n)})

# Slow loop
t0 = time.time()
result_loop = []
for v in big["val"]:
    result_loop.append(v * 2 + 1)
t_loop = time.time() - t0

# Fast vectorized
t0 = time.time()
result_vec = big["val"] * 2 + 1
t_vec = time.time() - t0

print(f"\nQ40 — Loop     : {t_loop*1000:.1f} ms")
print(f"      Vectorized: {t_vec*1000:.1f} ms")
print(f"      Speedup   : ~{t_loop/t_vec:.0f}×")
```

### ❓ Q44. Merge DataFrames using inner join.
### ❓ Q45. Merge using left join.

### ❓ Q47. Merge using outer join.
### ❓ Q48. Merge on multiple columns.
### ❓ Q49. Join DataFrames using index.
### ❓ Q50. Compare merge vs join behavior.

```python
emp = pd.DataFrame({"EmpID":[1,2,3,4,5],
                    "Name":["Alice","Bob","Charlie","Diana","Eve"],
                    "DeptID":[10,20,10,30,20]})
dept = pd.DataFrame({"DeptID":[10,20,40],
                     "DeptName":["Data Science","Machine Learning","DevOps"]})

print("Employees:\n", emp)
print("\nDepartments:\n", dept)

# Q44 — Inner
print("\nQ44 — INNER merge (only matching rows):")
print(pd.merge(emp, dept, on="DeptID", how="inner"))

# Q45 — Left
print("\nQ45 — LEFT merge (all employees kept):")
print(pd.merge(emp, dept, on="DeptID", how="left"))

# Q46 — Right
print("\nQ46 — RIGHT merge (all depts kept):")
print(pd.merge(emp, dept, on="DeptID", how="right"))

# Q47 — Outer
print("\nQ47 — OUTER merge (all rows from both):")
print(pd.merge(emp, dept, on="DeptID", how="outer"))

# Q48 — Multiple keys
sales  = pd.DataFrame({"Year":[2023,2023,2024],"Month":[1,2,1],"Sales":[100,150,200]})
target = pd.DataFrame({"Year":[2023,2023,2024],"Month":[1,2,1],"Target":[120,140,180]})
print("\nQ48 — Merge on multiple keys [Year, Month]:")
print(pd.merge(sales, target, on=["Year","Month"]))

# Q49 — join() (index-based)
df_left  = emp.set_index("EmpID")[["Name"]]
df_right = pd.DataFrame({"Level":["Junior","Senior","Junior"]},index=[1,2,3])
print("\nQ49 — join() on index:")
print(df_left.join(df_right, how="left"))

# Q50 — merge vs join comparison
print("\nQ50 — merge (column-based) vs join (index-based):")
print("  merge uses 'on' parameter → column names")
print("  join  uses index by default → set_index before joining")
print("\nSame result via merge:")
print(pd.merge(df_left.reset_index(), df_right.reset_index(),
               left_on="EmpID", right_on="index", how="left")[["EmpID","Name","Level"]])
```

### ❓ Q51. Group data using `groupby()`.
### ❓ Q52. Apply aggregation functions (sum, mean, etc.).
### ❓ Q53. Group by multiple columns.
### ❓ Q54. Filter groups.
### ❓ Q55. Use `transform()` in groupby.
### ❓ Q56. Apply custom functions in groupby.

```python
df = df_main.copy()

# Q51 — Basic groupby
print("Q51 — GroupBy Dept — mean Score:")
print(df.groupby("Dept")["Score"].mean())

# Q52 — Multiple aggregations
print("\nQ52 — Multiple agg on Score:")
print(df.groupby("Dept")["Score"].agg(["mean","min","max","std","count"]).round(2))

print("\nDifferent agg per column:")
print(df.groupby("Dept").agg(
    avg_score   =("Score",  "mean"),
    total_salary=("Salary", "sum"),
    headcount   =("Name",   "count")
))

# Q53 — Group by multiple columns
print("\nQ53 — GroupBy [Dept, City] — mean Score:")
print(df.groupby(["Dept","City"])["Score"].mean().reset_index())

# Q54 — filter(): keep groups where mean Score > 85
print("\nQ54 — filter groups where mean Score > 85:")
print(df.groupby("Dept").filter(lambda g: g["Score"].mean() > 85)[["Name","Dept","Score"]])

# Q55 — transform(): normalize score within dept (z-score)
df["Score_norm"] = df.groupby("Dept")["Score"].transform(
    lambda x: (x - x.mean()) / x.std()
)
print("\nQ55 — transform() z-score within Dept:")
print(df[["Name","Dept","Score","Score_norm"]].round(3))

# Q56 — Custom function
def score_range(s): return s.max() - s.min()
print("\nQ56 — Custom agg (score range per Dept):")
print(df.groupby("Dept")["Score"].agg(score_range))
```

## NP41–NP50: ML-Ready Dataset Preparation

```python
df = df_main.copy()

# NP41 — Random sampling with NumPy
np.random.seed(42)
sample_idx = np.random.choice(df.index, size=5, replace=False)
print("NP41 — Random sample (5 rows):")
print(df.loc[sample_idx][["Name","Score"]])

# NP42 — Shuffle data using NumPy
shuffled_idx = np.random.permutation(df.index)
df42 = df.loc[shuffled_idx].reset_index(drop=True)
print("\nNP42 — Shuffled DataFrame (first 5):")
print(df42[["Name","Score"]].head())

# NP43 — Percentile functions
arr43 = df["Score"].values
print("\nNP43 — Percentile analysis of Score:")
for p in [10, 25, 50, 75, 90, 95]:
    print(f"  P{p:2d}: {np.percentile(arr43, p):.2f}")

# NP44 — Variance and std
print("\nNP44 — Variance and Std of Score:")
print(f"  np.var : {np.var(arr43):.4f}")
print(f"  np.std : {np.std(arr43):.4f}")
print(f"  pd.var : {df['Score'].var():.4f}  (ddof=1, Bessel's correction)")
print(f"  pd.std : {df['Score'].std():.4f}  (ddof=1)")

# NP45 — NumPy slicing inside DataFrame
arr45 = df["Score"].values
print("\nNP45 — NumPy slicing:")
print("  arr[::2]  (every other):", arr45[::2])
print("  arr[::-1] (reversed):", arr45[::-1])
df["ReversedIdx"] = arr45[::-1]
print(df[["Name","Score","ReversedIdx"]].head())

# NP46 — Replace column with NumPy array
new_scores = np.round(np.clip(df["Score"].values * 1.05, 0, 100), 1)  # 5% bonus, max 100
df["Score_Adjusted"] = new_scores
print("\nNP46 — Replace column with NumPy array (5% bonus capped at 100):")
print(df[["Name","Score","Score_Adjusted"]])

# NP47 — Rolling computation using NumPy (manual)
scores = df["Score"].values
window = 3
rolling_mean_np = np.array([
    np.mean(scores[max(0,i-window+1):i+1]) for i in range(len(scores))
])
df["Rolling3_np"] = rolling_mean_np.round(2)
print("\nNP47 — Manual NumPy rolling mean (window=3):")
print(df[["Name","Score","Rolling3_np"]])

# NP48 — Feature Engineering Pipeline
print("\nNP48 — Feature Engineering Pipeline:")
df48 = df[["Score","Salary","Age","Exp"]].copy()

# Polynomial feature: Score²
df48["Score_sq"]   = np.square(df48["Score"])
# Log feature
df48["Log_Salary"] = np.log(df48["Salary"]).round(3)
# Interaction feature
df48["Score_Exp"]  = df48["Score"] * df48["Exp"]
# Bin feature
df48["Age_bin"]    = np.digitize(df48["Age"], bins=[25, 30, 35])
print(df48)

# NP49 — Build ML-ready dataset (features X, labels y)
print("\nNP49 — Build ML-ready dataset:")
df49 = df_main.copy()
df49["Label"] = (df49["Score"] >= 85).astype(int)   # binary target

# Encode Dept with np.select
dept_vals = df49["Dept"].values
dept_enc  = np.select(
    [dept_vals=="DS", dept_vals=="ML", dept_vals=="AI"],
    [0, 1, 2], default=-1
)
df49["Dept_enc"] = dept_enc

X = df49[["Age","Score","Salary","Exp","Dept_enc"]].to_numpy()
y = df49["Label"].to_numpy()
print(f"  X shape: {X.shape}, y shape: {y.shape}")
print(f"  X dtype: {X.dtype}, y dtype: {y.dtype}")
print(f"  Label distribution: {np.bincount(y)}")
```

```python
# NP50 — End-to-End Analysis: NumPy + Pandas Pipeline
print("=" * 65)
print("NP50 — END-TO-END NUMPY + PANDAS ANALYSIS PIPELINE")
print("=" * 65)

np.random.seed(99)
n = 200

# Step 1: Generate raw dataset with NumPy
raw = pd.DataFrame({
    "EmpID"   : np.arange(1, n+1),
    "Age"     : np.random.randint(22, 60, n),
    "Exp"     : np.random.randint(0, 20, n),
    "Score"   : np.round(np.random.normal(75, 12, n), 1),
    "Salary"  : np.random.randint(40000, 130000, n),
    "DeptCode": np.random.choice([0, 1, 2], n),    # 0=DS, 1=ML, 2=AI
})
print(f"\n📥 Step 1: Generated {n} rows | Shape: {raw.shape}")

# Step 2: Clean — clip invalid scores
raw["Score"] = np.clip(raw["Score"], 0, 100).round(1)
print(f"\n🧹 Step 2: Score clipped to [0,100]")

# Step 3: Feature Engineering
raw["Log_Salary"]  = np.log(raw["Salary"]).round(3)
raw["Score_sq"]    = np.square(raw["Score"]).round(1)
raw["Score_Exp"]   = raw["Score"] * raw["Exp"]
raw["Normalized_Score"] = ((raw["Score"] - raw["Score"].min()) /
                           (raw["Score"].max() - raw["Score"].min())).round(3)
raw["Label"] = np.where(raw["Score"] >= 80, 1, 0)
print(f"\n⚙️  Step 3: Added 5 engineered features")

# Step 4: Statistical analysis by dept
dept_names = {0:"DS", 1:"ML", 2:"AI"}
raw["Dept"] = raw["DeptCode"].map(dept_names)
summary = raw.groupby("Dept").agg(
    count   =("EmpID",  "count"),
    avg_age =("Age",    "mean"),
    avg_score=("Score", "mean"),
    avg_sal  =("Salary","mean"),
    label_pct=("Label", "mean")
).round(2)
print(f"\n📊 Step 4: Department Summary:")
print(summary)

# Step 5: Outlier removal
q1, q3 = np.percentile(raw["Score"], [25,75])
iqr = q3 - q1
raw_clean = raw[(raw["Score"] >= q1-1.5*iqr) & (raw["Score"] <= q3+1.5*iqr)]
print(f"\n🚫 Step 5: Removed {len(raw)-len(raw_clean)} outliers | Remaining: {len(raw_clean)}")

# Step 6: ML feature matrix
feature_cols = ["Age","Exp","Score","Salary","Log_Salary","Score_sq","Normalized_Score"]
X = raw_clean[feature_cols].to_numpy(dtype=np.float32)
y = raw_clean["Label"].to_numpy()
print(f"\n🤖 Step 6: ML Feature matrix X: {X.shape}, Labels y: {y.shape}")
print(f"   Label balance: 0={np.sum(y==0)}, 1={np.sum(y==1)}")
print(f"\n✅ Pipeline complete — data ready for ML model training!")
```
